## Este es un ejemplo de SVM con la base de datos IRIS

### STEP 1: Understand the Business

<img src="happy.jpg style="width:200px;length:200px"/>

```python
#Import the libraries and data sets
from sklearn import svm
from sklearn import datasets
```

```python
#Load the data set
iris = datasets.load_iris()
```

```python
#explore loaded dataset
type(iris)  #see evenchrything as a bunch
```

```python
iris.feature_names  #the columns names
```

```python
iris.target  #target is what we are going to predict 0==setosa, 1==virginica, 2==versicolor 
iris.target_names  #see the names of the targets
```

```python
#Create different arrays to store the dependent and independent variables
x = iris.data[:,2]  #use all the rows, and the first two columns (features)
y = iris.target #Y will take the predicted value (type of iris flower)
```

### STEP 4: Model

```python
#import the module that will allow splitting the data
from sklearn.model_selection import train_test_split 
```

```python
#Split the data into two sets: training data and the validaton data
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=4) 
```

```python
#reshape the data
x_train_mod = x_train.reshape(-1,1)
x_test_mod = x_test.reshape(-1,1)
y_train_mod = y_train.reshape(-1,1)
y_test_mod = y_test.reshape(-1,1)
```

```python
#Create a SVM model and train it
model = svm.SVC(kernel='linear') #creates a new feature to create the hyperplane that separates the data into 2 classes
model.fit(x_train_mod, y_train_mod)
```

### STEP 5: TEST Performance

```python
#Now use the model to predict
y_pred_mod = model.predict(x_test_mod)
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test_mod, y_pred_mod))
```
