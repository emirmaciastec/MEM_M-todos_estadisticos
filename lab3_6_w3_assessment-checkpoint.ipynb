## Scatter plots

First, let's looks at two variables that we expect to have a strong relationship, 'Height' and 'Wingspan'.

````python
# Make a Seaborn scatter plot with x = height and y = wingspan using sns.scatterplot(x, y)
sns.scatterplot(x=df['Height'], y=df['Wingspan'])
plt.show()
````

*   **Are there outliers?** There don't appear to be any significant outliers that deviate drastically from the general trend.
*   **Are their ranges similar or different?** The ranges appear to be similar. Both variables are measured in inches and have a comparable spread of values.
*   **How else could you describe the relationship?** There is a strong positive correlation between height and wingspan.
